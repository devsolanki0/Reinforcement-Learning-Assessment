1 – Exact Prompt Given to AI

Write a simple Python program that simulates a food delivery reinforcement learning episode. The program should have an agent that takes a sequence of actions and accumulates rewards. It should track the agent's location, total reward, and step count, update the state after every action, print a per-step log containing the action, reward, and running total reward, and print a final episode summary containing total reward and total steps.


```python
# 1

class DeliveryAgent:
    def __init__(self):
        self.location = "Base"
        self.total_reward = 0
        self.step_count = 0

    def take_action(self, action):
        rewards = {
            "accept_order": 2,
            "navigate": 1,
            "deliver": 10,
            "wait": -1
        }

        reward = rewards.get(action, 0)
        self.total_reward += reward
        self.step_count += 1

        if action == "accept_order":
            self.location = "Restaurant"
        elif action == "navigate":
            self.location = "Customer"
        elif action == "deliver":
            self.location = "Base"
        else:
            self.location = "Waiting Area"

        print(
            f"Step {self.step_count}: "
            f"Action={action}, Reward={reward}, "
            f"Total Reward={self.total_reward}"
        )

```

```python
actions = [
    "accept_order",
    "navigate",
    "deliver",
    "wait",
    "accept_order"
]

agent = DeliveryAgent()

for action in actions:
    agent.take_action(action)

print("Final Episode Summary")
print("Total Reward:", agent.total_reward)
print("Total Steps:", agent.step_count)
```

```python
# 2

class DeliveryAgent:
    def __init__(self):
        self.location = "Base"
        self.total_reward = 0
        self.step_count = 0

    def take_action(self, action):
        rewards = {
            "accept_order": 2,
            "navigate": 1,
            "deliver": 10,
            "wait": -1
        }

        if action not in rewards:
            raise ValueError(f"Invalid action: {action}")

        reward = rewards[action]
        self.total_reward += reward
        self.step_count += 1

        if action == "accept_order":
            self.location = "Restaurant"
        elif action == "navigate":
            self.location = "Customer"
        elif action == "deliver":
            self.location = "Base"
        elif action == "wait":
            self.location = "Waiting Area"

        print(
            f"Step {self.step_count}: "
            f"Action={action}, "
            f"Location={self.location}, "
            f"Reward={reward}, "
            f"Running Total={self.total_reward}"
        )

```

```python
actions = [
    "accept_order",
    "navigate",
    "deliver",
    "wait",
    "accept_order"
]

agent = DeliveryAgent()

for action in actions:
    agent.take_action(action)

print("\nFinal Episode Summary")
print("Total Reward:", agent.total_reward)
print("Total Steps:", agent.step_count)
print("Final Location:", agent.location)
```
