```python
class FoodDeliveryRLSimulator:
    def __init__(self):
        self.agent_state = {
            "current_location": "Base",
            "orders_delivered": 0,
            "total_reward": 0
        }

        self.total_episodes = 0
        self.total_reward_all_episodes = 0

    def reward(self, action):
        rewards = {
            "accept_order": 2,
            "navigate_to_zone": 2,
            "request_directions": 0,
            "mark_delivered": 10,
            "wait": -1
        }
        return rewards.get(action, -2)

    def update_state(self, action):
        locations = {
            "accept_order": "Restaurant",
            "navigate_to_zone": "Delivery Zone",
            "request_directions": "On Route",
            "mark_delivered": "Customer",
            "wait": "Waiting Area"
        }

        self.agent_state["current_location"] = locations[action]

        if action == "mark_delivered":
            self.agent_state["orders_delivered"] += 1

    def start_episode(self):
        actions = [
            "accept_order",
            "navigate_to_zone",
            "request_directions",
            "mark_delivered",
            "wait"
        ]

        episode_reward = 0

        print("\n===== NEW DELIVERY EPISODE =====")

        for step, action in enumerate(actions, 1):
            reward = self.reward(action)
            self.update_state(action)

            self.agent_state["total_reward"] += reward
            episode_reward += reward

            print(
                f"Step {step}: {action} | "
                f"Reward: {reward} | "
                f"Running Reward: {episode_reward}"
            )
            print("State:", self.agent_state)

        self.total_episodes += 1
        self.total_reward_all_episodes += episode_reward

        print("\n----- Episode Summary -----")
        print("Total Episode Reward:", episode_reward)
        print("Total Steps:", len(actions))

    def view_stats(self):
        if self.total_episodes:
            average = self.total_reward_all_episodes / self.total_episodes
        else:
            average = 0

        print("\n===== AGENT STATS =====")
        print("Total Episodes Run:", self.total_episodes)
        print("Total Reward Earned:", self.total_reward_all_episodes)
        print("Average Reward per Episode:", round(average, 2))
        print("Current Agent State:", self.agent_state)
```

Run the Simulator

```python
simulator = FoodDeliveryRLSimulator()

simulator.start_episode()
simulator.start_episode()

simulator.view_stats()
```
