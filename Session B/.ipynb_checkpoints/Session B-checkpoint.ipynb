```python
# 1

def calculate_reward(delivery_time, is_on_time, customer_rating):
    reward = 10 if is_on_time else -5
    reward += customer_rating

    print(
        f"Delivery time: {delivery_time} minutes | "
        f"On time: {is_on_time} | "
        f"Customer rating: {customer_rating} | "
        f"Final reward: {reward}"
    )
    return reward


calculate_reward(25, True, 5)
calculate_reward(40, False, 2)
calculate_reward(30, True, 3)
```

```python
# 2

state = {
    "location": "Restaurant",
    "orders_delivered": 0,
    "total_reward": 0.0,
    "is_available": True
}


def update_state(state, new_location, reward_earned):
    state["location"] = new_location
    state["orders_delivered"] += 1
    state["total_reward"] += reward_earned


transitions = [
    ("Zone A", 12),
    ("Zone B", 8),
    ("Zone C", 15),
    ("Zone D", 10)
]

for location, reward in transitions:
    update_state(state, location, reward)
    print("Current state:", state)
```

```python
# 3

actions = [
    "accept_order",
    "request_directions",
    "mark_delivered",
    "accept_order",
    "reject_order",
    "mark_delivered"
]

action_rewards = {
    "accept_order": 2,
    "reject_order": -1,
    "request_directions": 0,
    "mark_delivered": 10
}

cumulative_reward = 0

for step, action in enumerate(actions, 1):
    reward = action_rewards[action]
    cumulative_reward += reward

    print(
        f"Step {step}: Action = {action}, "
        f"Reward = {reward}, "
        f"Cumulative Reward = {cumulative_reward}"
    )

print("Total Episode Reward:", cumulative_reward)
```

```python
# 4

class DeliveryEnvironment:
    def __init__(self):
        self.current_state = {
            "location": "Base",
            "pending_orders": 3
        }
        self.total_reward = 0
        self.step_count = 0

    def step(self, action):
        self.step_count += 1

        if action == "deliver_order":
            if self.current_state["pending_orders"] > 0:
                self.current_state["pending_orders"] -= 1
                self.current_state["location"] = "Customer"
                reward = 10
            else:
                reward = 0

        elif action == "wait":
            self.current_state["location"] = "Waiting Area"
            reward = -1

        elif action == "navigate_to_zone":
            self.current_state["location"] = "Delivery Zone"
            reward = 2

        else:
            reward = -2

        self.total_reward += reward

        print(f"Step {self.step_count}: {action}")
        print("New state:", self.current_state)
        print("Reward:", reward)
        print("Total reward:", self.total_reward)
        print("-" * 40)

        return reward


env = DeliveryEnvironment()

for action in [
    "navigate_to_zone",
    "deliver_order",
    "navigate_to_zone",
    "deliver_order",
    "wait",
    "deliver_order"
]:
    env.step(action)

print("Final Total Reward:", env.total_reward)
print("Total Steps:", env.step_count)
```
